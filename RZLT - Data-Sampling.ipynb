```python
import geopandas as gpd

gdf_full = gpd.read_file("Datasets/RZLT-Dataset.geojson")
```

```python
# Check value distributions across both potential columns
print("LOCAL_AUTHORITY counts:")
print(gdf_full['LOCAL_AUTHORITY'].value_counts(dropna=False))

print("\nLOCAL_AUTHORITY_NAME counts:")
print(gdf_full['LOCAL_AUTHORITY_NAME'].value_counts(dropna=False))
```

```python
# Strip leading/trailing whitespace and convert to upper
gdf_full['LOCAL_AUTHORITY_CLEAN'] = gdf_full['LOCAL_AUTHORITY'].astype(str).str.strip().str.upper()

# Filter using clean code OR full name
gdf_sample = gdf_full[
    (gdf_full['LOCAL_AUTHORITY_CLEAN'] == 'CWLA') | 
    (gdf_full['LOCAL_AUTHORITY_NAME'].str.contains('Carlow', case=False, na=False))
].copy()

print(f"Filtered records count: {len(gdf_sample)}")
```

```python
# Filter for a specific local authority
gdf_sample.to_file("Datasets/RZLT_Samples/rzlt_sample_CWLA.geojson")

print("GDF Full:", gdf_full['LIVE_SITE'].value_counts())
print("GDF Sample:", gdf_sample['LIVE_SITE'].value_counts())
```
